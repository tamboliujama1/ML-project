```python
# ============================================================
#       EDUTRACK: STUDENT PERFORMANCE PREDICTION SYSTEM
#       Machine Learning Mini Project
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

# ============================================================
# 1. PROJECT INTRODUCTION
# ============================================================

print("=" * 65)
print("       EDUTRACK - STUDENT PERFORMANCE PREDICTION")
print("=" * 65)

print("\nProject Objectives:")
print("1. Predict students' final grades using Linear Regression.")
print("2. Identify students who may be at risk of failing.")
print("3. Compare Logistic Regression and AdaBoost.")
print("4. Group students using K-Means Clustering.")
print("=" * 65)


# ============================================================
# 2. LOAD DATASET
# ============================================================

print("\n[1] UPLOAD DATASET")

filename = "student-mat.csv"
df = pd.read_csv(filename, sep=";")

print("\nDataset loaded successfully!")
print("Total Students:", df.shape[0])
print("Total Columns:", df.shape[1])

print("\nFirst Five Records:")
display(df.head())

print("\nDataset Information:")
print("Missing Values:", df.isnull().sum().sum())

print("\nStatistical Summary:")
display(df.describe())


# ============================================================
# 3. DATA ANALYSIS
# ============================================================

print("\n[2] DATA ANALYSIS")

print("\nFinal Grade Statistics:")
print(df["G3"].describe())

print("\nAverage Final Grade:", round(df["G3"].mean(), 2))

print("\nStudents Passing:", int((df["G3"] >= 10).sum()))
print("Students At Risk:", int((df["G3"] < 10).sum()))

# Grade distribution
plt.figure(figsize=(8, 5))
plt.hist(df["G3"], bins=15, edgecolor="black")
plt.title("Distribution of Students' Final Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")
plt.grid(axis="y", alpha=0.3)
plt.show()


# ============================================================
# 4. PREPARE DATA
# ============================================================

print("\n[3] PREPARING DATA")

# G3 is the final grade that we want to predict.
# Remove G3 from the input features to avoid target leakage.

X = df.drop("G3", axis=1)
y = df["G3"]

# Convert categorical columns into numerical columns.
X = pd.get_dummies(X, drop_first=True, dtype=int)

# Binary classification:
# 0 = At Risk (G3 < 10)
# 1 = Likely to Pass (G3 >= 10)

y_class = (df["G3"] >= 10).astype(int)

# Split the data into training and testing sets.
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

# Use the same split for classification.
y_class_train = y_class.loc[X_train.index]
y_class_test = y_class.loc[X_test.index]

print("Training Records:", len(X_train))
print("Testing Records:", len(X_test))
print("Input Features:", X.shape[1])


# ============================================================
# 5. MULTIPLE LINEAR REGRESSION
# ============================================================

print("\n" + "=" * 65)
print("[4] MULTIPLE LINEAR REGRESSION")
print("=" * 65)

linear_model = LinearRegression()
linear_model.fit(X_train, y_train)

linear_predictions = linear_model.predict(X_test)

mae = mean_absolute_error(y_test, linear_predictions)
mse = mean_squared_error(y_test, linear_predictions)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, linear_predictions)

print("\nModel Results:")
print("Mean Absolute Error (MAE):", round(mae, 2))
print("Mean Squared Error (MSE):", round(mse, 2))
print("Root Mean Squared Error (RMSE):", round(rmse, 2))
print("R2 Score:", round(r2, 3))

# Compare actual and predicted grades.
grade_comparison = pd.DataFrame({
    "Actual Grade": y_test.values,
    "Predicted Grade": np.round(linear_predictions, 2)
})

print("\nActual vs Predicted Grades:")
display(grade_comparison.head(10))

# Graph: actual vs predicted grades
plt.figure(figsize=(7, 5))
plt.scatter(y_test, linear_predictions, alpha=0.7)
plt.plot([0, 20], [0, 20], linestyle="--")
plt.title("Actual vs Predicted Final Grades")
plt.xlabel("Actual Grade")
plt.ylabel("Predicted Grade")
plt.grid(alpha=0.3)
plt.show()


# ============================================================
# 6. LOGISTIC REGRESSION
# ============================================================

print("\n" + "=" * 65)
print("[5] LOGISTIC REGRESSION")
print("=" * 65)

# Scale the features for Logistic Regression.
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

logistic_model = LogisticRegression(max_iter=2000, random_state=42)

logistic_model.fit(X_train_scaled, y_class_train)

logistic_predictions = logistic_model.predict(X_test_scaled)

logistic_accuracy = accuracy_score(
    y_class_test, logistic_predictions
)

print("\nModel Accuracy:", round(logistic_accuracy * 100, 2), "%")

print("\nClassification Report:")
print(
    classification_report(
        y_class_test,
        logistic_predictions,
        labels=[0, 1],
        target_names=["At Risk", "Likely to Pass"],
        zero_division=0
    )
)

print("Confusion Matrix:")
print(confusion_matrix(y_class_test, logistic_predictions))

ConfusionMatrixDisplay.from_predictions(
    y_class_test,
    logistic_predictions,
    labels=[0, 1],
    display_labels=["At Risk", "Likely to Pass"],
    cmap="Blues"
)

plt.title("Logistic Regression - Confusion Matrix")
plt.show()


# ============================================================
# 7. ADABOOST CLASSIFIER
# ============================================================

print("\n" + "=" * 65)
print("[6] ADABOOST CLASSIFIER")
print("=" * 65)

# AdaBoost combines several weak learners.
# Decision stumps are shallow decision trees.

adaboost_model = AdaBoostClassifier(
    DecisionTreeClassifier(max_depth=1, random_state=42),
    n_estimators=100,
    learning_rate=0.8,
    random_state=42
)

adaboost_model.fit(X_train, y_class_train)

adaboost_predictions = adaboost_model.predict(X_test)

adaboost_accuracy = accuracy_score(
    y_class_test, adaboost_predictions
)

print("\nAdaBoost Accuracy:", round(adaboost_accuracy * 100, 2), "%")

print("\nClassification Report:")
print(
    classification_report(
        y_class_test,
        adaboost_predictions,
        labels=[0, 1],
        target_names=["At Risk", "Likely to Pass"],
        zero_division=0
    )
)

print("Confusion Matrix:")
print(confusion_matrix(y_class_test, adaboost_predictions))

ConfusionMatrixDisplay.from_predictions(
    y_class_test,
    adaboost_predictions,
    labels=[0, 1],
    display_labels=["At Risk", "Likely to Pass"],
    cmap="Oranges"
)

plt.title("AdaBoost - Confusion Matrix")
plt.show()


# ============================================================
# 8. COMPARE CLASSIFICATION MODELS
# ============================================================

print("\n" + "=" * 65)
print("[7] MODEL COMPARISON")
print("=" * 65)

comparison = pd.DataFrame({
    "Model": ["Logistic Regression", "AdaBoost"],
    "Accuracy (%)": [
        round(logistic_accuracy * 100, 2),
        round(adaboost_accuracy * 100, 2)
    ]
})

display(comparison)

plt.figure(figsize=(8, 5))
plt.bar(comparison["Model"], comparison["Accuracy (%)"])
plt.title("Classification Model Accuracy Comparison")
plt.ylabel("Accuracy (%)")
plt.ylim(0, 100)

for i, value in enumerate(comparison["Accuracy (%)"]):
    plt.text(i, value + 1, f"{value:.2f}%", ha="center")

plt.show()


# ============================================================
# 9. K-MEANS CLUSTERING
# ============================================================

print("\n" + "=" * 65)
print("[8] K-MEANS CLUSTERING")
print("=" * 65)

# Use input features only; G3 is not used to form clusters.
cluster_scaler = StandardScaler()
X_cluster_scaled = cluster_scaler.fit_transform(X)

# Divide students into three groups.
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)

df["Cluster"] = kmeans.fit_predict(X_cluster_scaled)

# Summarize clusters.
cluster_summary = df.groupby("Cluster").agg(
    Student_Count=("G3", "count"),
    Average_Final_Grade=("G3", "mean"),
    Average_Study_Time=("studytime", "mean"),
    Average_Absences=("absences", "mean")
).round(2)

print("\nCluster Summary:")
display(cluster_summary)

# Visualize clusters using two features.
# These two features are selected only for visualization.
plt.figure(figsize=(8, 5))

for cluster in sorted(df["Cluster"].unique()):
    cluster_data = df[df["Cluster"] == cluster]

    plt.scatter(
        cluster_data["studytime"],
        cluster_data["absences"],
        label=f"Cluster {cluster}",
        alpha=0.7
    )

plt.title("Student Groups Using K-Means")
plt.xlabel("Study Time")
plt.ylabel("Absences")
plt.legend()
plt.grid(alpha=0.3)
plt.show()


# ============================================================
# 10. SAMPLE STUDENT PREDICTION
# ============================================================

print("\n" + "=" * 65)
print("[9] SAMPLE STUDENT PREDICTION")
print("=" * 65)

# Select one student from the test dataset.
sample_student = X_test.iloc[[0]]

# Predict the student's final grade.
predicted_grade = linear_model.predict(sample_student)[0]

# Predict pass/at-risk category using Logistic Regression.
sample_scaled = scaler.transform(sample_student)
predicted_class = logistic_model.predict(sample_scaled)[0]

print("\nSample Student Information:")
print("Study Time:", df.loc[sample_student.index[0], "studytime"])
print("Previous Grade G1:", df.loc[sample_student.index[0], "G1"])
print("Previous Grade G2:", df.loc[sample_student.index[0], "G2"])
print("Absences:", df.loc[sample_student.index[0], "absences"])

print("\nPrediction Results:")
print("Predicted Final Grade:", round(predicted_grade, 2), "/ 20")

if predicted_class == 0:
    print("Status: AT RISK")
    print("Suggestion: Consider additional academic support.")
else:
    print("Status: LIKELY TO PASS")
    print("Suggestion: Continue the current study routine.")


# ============================================================
# 11. FINAL PROJECT SUMMARY
# ============================================================

print("\n" + "=" * 65)
print("                 PROJECT SUMMARY")
print("=" * 65)

print("1. Linear Regression predicted students' final grades.")
print("2. Logistic Regression classified students by pass risk.")
print("3. AdaBoost provided a second classification model.")
print("4. K-Means grouped students using their input features.")

print("\nRegression Results:")
print("MAE:", round(mae, 2))
print("RMSE:", round(rmse, 2))
print("R2 Score:", round(r2, 3))

print("\nClassification Results:")
print("Logistic Regression:", round(logistic_accuracy * 100, 2), "%")
print("AdaBoost:", round(adaboost_accuracy * 100, 2), "%")

print("\nDataset Records:", len(df))
print("Number of Clusters:", df["Cluster"].nunique())

print("\nNOTE:")
print("These predictions are educational estimates, not definitive")
print("judgments of a student's ability or future performance.")

print("\n" + "=" * 65)
print("             EDUTRACK EXECUTION COMPLETED")
print("=" * 65)
```
